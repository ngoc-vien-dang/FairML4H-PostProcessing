```python
import pandas as pd
import numpy as np
from google.colab import drive
import os

# Mount Google Drive
drive.mount('/content/drive')
import pickle
# Save the filtered_dfs dictionary
file_path = '/content/drive/My Drive/2024/Project_5/Results/Add_XGB_DNN_Trade-off_overall_data.pkl'
# Load the filtered_dfs dictionary
with open(file_path, 'rb') as f:
    overall_df = pickle.load(f)
```

```python
# Define the method replacements
method_replacements = {
    'cpp': 'CPP',
    'epp': 'EPP',
    'fact': 'FACT',
    'gstar': 'GSTAR',
    'multiaccboost': 'MAB',
    'psta': 'PSTA',
    'roc': 'ROC',
    'All': 'Average'
}

# Define the desired order
method_order = ['CPP', 'EPP', 'ROC', 'PSTA', 'FACT', 'GSTAR', 'MAB', 'Average']

# Apply filtering, renaming, and sorting to all keys in overall_df
for key in overall_df.keys():
    # Filter out 'baseline'
    overall_df[key] = overall_df[key].query("method != 'baseline'").copy()

    # Rename methods
    overall_df[key]['method'] = overall_df[key]['method'].replace(method_replacements)

    # Convert 'method' column to categorical with the specified order
    overall_df[key]['method'] = pd.Categorical(overall_df[key]['method'], categories=method_order, ordered=True)

    # Sort by the new categorical order
    overall_df[key] = overall_df[key].sort_values('method')
```
